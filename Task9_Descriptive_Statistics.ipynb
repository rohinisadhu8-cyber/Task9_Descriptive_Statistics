```python
import pandas as pd
from sklearn.datasets import load_iris

# Load the Iris dataset
iris = load_iris(as_frame=True)
df = iris.frame

# Display the first five rows
print(df.head())
```

```python
print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nMissing values:")
print(df.isnull().sum())
```

```python
# Select the four numeric flower measurements
numeric_df = df[iris.feature_names]

# Calculate mean, median, mode, standard deviation
# and percentiles
summary = pd.DataFrame({
    "Mean": numeric_df.mean(),
    "Median": numeric_df.median(),
    "Mode": numeric_df.mode().iloc[0],
    "Standard Deviation": numeric_df.std(),
    "25th Percentile": numeric_df.quantile(0.25),
    "50th Percentile": numeric_df.quantile(0.50),
    "75th Percentile": numeric_df.quantile(0.75)
})

summary = summary.round(3)
display(summary)
```

Interpretation of Descriptive Statistics

Descriptive statistics were calculated for the four numerical features of the Iris dataset: sepal length, sepal width, petal length, and petal width.

The mean and median help identify the central values of each feature. Comparing them helps us understand whether the data may be skewed. The mode identifies the most frequently occurring value.

Standard deviation measures how much the measurements vary. A higher standard deviation indicates greater variation in the values. The 25th and 75th percentiles help describe the spread of the middle 50% of the observations.

Overall, descriptive statistics provide a simple and useful summary of the dataset and help us understand its distribution and variability.
